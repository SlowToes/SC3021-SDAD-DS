
# **DS2**

Instagram Reel dataset (DS2) - [Instagram-Reels.csv](https://brightdata.com/cp/datasets/browse/gd_lyclm20il4r5helnj?id=hl_6c69a494)

Strengths:

* Directly supports video-specific analysis with complete length (duration in seconds, mean 60s, no missing values) and strong engagement metrics like views (mean ~73k, up to 25M) and video_play_count.
​

* Full coverage of core fields like date_posted, likes, num_comments, enabling easy extraction of posting time and hashtag counts (from hashtags, 65.6% non-null).
​

* Includes creator context (followers, is_verified) for normalizing virality by audience size.
​

Weaknesses:

* High missingness in followers (48.1%) and posts_count (48.1%), limiting audience-adjusted models unless imputed.
​

* Data quality issues like negative likes values and extreme outliers (e.g., views std ~826k), requiring cleaning and log-transformation.
​

* No explicit "trending audio" flag; audio_url needs frequency-based proxy.

```python
df = pd.read_csv('Instagram - Reels.csv')
```

```python
df.info()
```

```python
df.sort_values(by='likes', ascending=False).head(10)
```

```python
df.describe(include="all")
```

# **DS3**

Instagram generic dataset (DS3) - [Instagram-datasets.csv](https://github.com/luminati-io/Social-media-dataset-samples/blob/main/Instagram-datasets.csvhttps://github.com/luminati-io/Social-media-dataset-samples/blob/main/Instagram-datasets.csv)

Strengths:


*   Near-perfect completeness (0% missing in core columns like likes_number, replies_number, comment_date, post_id), ideal for aggregating comment engagement per post.

*  Enables label enrichment (e.g., comments as virality signal) via post_id joins.


* Low skew in metrics (e.g., likes_number mean 3.7), stable for summaries.


Weaknesses:

* Lacks post characteristics (no duration, hashtags, audio), only useful as an extension.
​

* High missingness in hashtag_comment (99.4%) and tagged_users_in_comment (85.4%), limiting content analysis from comments.
​

* Potential selection bias: only commented posts, skewing toward higher-engagement content.


```python
df = pd.read_csv('Instagram-datasets.csv')

```
