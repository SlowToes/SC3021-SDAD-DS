# **Clearly state the data science question your team aims to answer.**

What is the relationship between specific post characteristics and the likelihood of a post achieving viral status on Social Media?

# **Identify key terms and provide the definition.**

* Post characteristics
    * We will examine multiple types of posts, but most of which should fall under videos and texts. 
    * Characteristics for videos include the duration, or even frame rate and video resolution if applicable.
    * Characteristics for texts include text length and the sentiment(tone) of the score.
    * Generic characteristics applicable to both formats include posting time (hour of day), day of week, number of hashtags, and engagement-related data such as likes, comments and shares.

* Viral status
    * We define a viral post as follows:
        * Speed of outreach: It gains popularity very fast, with views, shares, or mentions growing in an exponentially rather than slowly over time.
        * High visibility: It reaches a large audience relative to a normal post, and may appear beyond the post's original intended community.
    * If there is no data that shows the rate of post spread, i.e. the various timings for the post's growth of the number of shares, likes and comments, then we will use a preset time frame or preset number of shares/likes/comments to determine its virality.

* Social Media
    * For now, we are looking at TikTok, Facebook and Instagram, which are in the top 5 of most popular social platforms ([Wikipedia - List of most popular social platforms](https://en.wikipedia.org/wiki/List_of_most_popular_social_platforms))

# **Ask**

* **Problem to be solved:** Identify the key factors that contribute to the virality of a social media post.

* **Underlying kind of problem:**  Predictive task to estimate post virality based on different post characteristics.

* **Stackholders expectations:** Actionable insights that can aid in the creation of posts with a higher likelihood of being viral.

* **Goal of analysis:** Identifying key post characteristics that significantly influences post virality may guide the definition of features and content for a post.

* **Relevance to stackholders:** Enable stakeholders to strategically incorporate high-impact elements into their posts to increase the probability of making it viral.

# **Hypothesis Evaluation**

* **Null Hypothesis (H₀)**
    * Post characteristics have no statistically significant effect on whether a post achieves viral status.

* **Alternative Hypothesis (H₁)**
    * At least one post characteristic has a statistically significant effect on whether a post achieves viral status.

# **Discuss the data requirements, i.e., what data is needed to address the problem.**

**1. Post Metadata**
* Platform (e.g., Facebook/Meta, Instagram, Twitter/X, TikTok)
* Date and time of posting
* Day of week
* Account
* Number of followers at posting time

**2. Engagement Metrics**
* Number of likes/reactions
* Number of shares/retweets
* Number of comments
* Reach / impressions (if available)
* Engagement rate (engagement ÷ followers)
* Growth rate of engagement (e.g., engagement in first 1–24 hours) (if available)

**3. Content Characteristics**
* Content Type
    * Text-only, image, video etc
    * Video length
    * Text length
    * Image count
* Textual Features
    * Caption length
    * Hashtag count
    * Mentions/tags
    * Presence of emojis
    * Sentiment score (positive / negative / neutral)
    * Readability score
* Visual Features (if available)
    * Colorfulness / brightness
    * Presence of faces or people
    * Thumbnail quality (for videos)
* Audio information (if available)
    * Sound effects

**4. Others**
* Topic category of post
* Account used to post


# **DS5**

Instagram Reel dataset (DS2) - [Instagram-Reels.csv](https://brightdata.com/cp/datasets/browse/gd_lyclm20il4r5helnj?id=hl_6c69a494)

Strengths:

* Directly supports video-specific analysis with complete length (duration in seconds, mean 60s, no missing values) and strong engagement metrics like views (mean ~73k, up to 25M) and video_play_count.
​

* Full coverage of core fields like date_posted, likes, num_comments, enabling easy extraction of posting time and hashtag counts (from hashtags, 65.6% non-null).
​

* Includes creator context (followers, is_verified) for normalizing virality by audience size.
​

Weaknesses:

* High missingness in followers (48.1%) and posts_count (48.1%), limiting audience-adjusted models unless imputed.
​

* Data quality issues like negative likes values and extreme outliers (e.g., views std ~826k), requiring cleaning and log-transformation.
​

* No explicit "trending audio" flag; audio_url needs frequency-based proxy.

```python
df = pd.read_csv('Instagram - Reels.csv')
```

```python
df.info()
```

```python
df.sort_values(by='likes', ascending=False).head(10)
```
