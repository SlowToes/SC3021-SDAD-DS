## DS2

[Shofo TikTok General (Small) – engagement metrics (DS2) – tiktok_engagement_small.csv](https://huggingface.co/datasets/Shofo/shofo-tiktok-general-small)

This dataset contains engagement metrics for TikTok videos (e.g., play/view count, like count, comment count, share count).
These fields are used to approximate virality.

Strengths:
* Contains direct engagement measures which are usable to define viral vs non-viral.
* Allows ranking and threshold-based labeling (e.g., viral if play_count ≥ 1,000,000).

Weaknesses:
* “Virality” is not a universal definition thus thresholds are arbitary.


```python
eng = pd.read_csv("tiktok_engagement_small.csv")
eng.info()
```

## DS3

[Shofo TikTok General (Small) – text features (DS3) – tiktok_text.csv](https://huggingface.co/datasets/Shofo/shofo-tiktok-general-small)

This dataset contains text fields such as the caption/description and hashtags.

Strengths:
* Captions and hashtags can be quantified and linked to engagement.
* Useful for feature engineering without needing to download raw video.

Weaknesses:
* Text fields may be missing, noisy, or multilingual.
* Hashtags may be stored as strings and require standardization.

```python
txt = pd.read_csv("tiktok_text.csv")
txt.info()
```

```python
txt.head()
```

```python
txt.describe(include="all")
```

## Merge datasets (DS1 + DS2 + DS3)

We merge on video_id to combine post metadata, engagement metrics, and text fields into one analysis table.

```python
df = posts.merge(eng, on="video_id", how="inner").merge(txt, on="video_id", how="left")
df.info()
```
