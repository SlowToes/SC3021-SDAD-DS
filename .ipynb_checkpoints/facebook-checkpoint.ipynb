## DS1

[Facebook Text Classification (DS1) - Facebook_data_txt.csv](https://www.kaggle.com/datasets/kalashnikov1405/facebook-text-classification)  

The dataset contains Facebook post data, in particular text classification.

(Definition of text classification:
It is a labeled collection of text data (sentences or documents) paired with predefined categories, used to train machine learning models to automatically categorize future, unseen text.)

Strengths:
* Has many useful attributes that directly measures how viral a post is as well as the post's corresponding properties. For example, the timestamp, engagement metrics and sentiment will be helpful here.
* There may be potential for modelling of content popularity for prediction by using the likes, comments, shares data.
* Useful to identify text to be used in posts that can give the most chance of making it viral.

Weaknesses:
* Does not have any information related to videos, as it only focuses on text-based media.

```python
import pandas as pd
```

```python
df = pd.read_csv('Facebook_data_txt.csv')
```

```python
df.info()
```

```python
# change the by to likes,comments,shares etc
df.sort_values(by='likes', ascending=False).head(10)
```

```python
df.describe(include="all")
```

## DS2

[Facebook posts dataset Bright Data (DS2) - facebook-posts-by-profile.csv](https://github.com/luminati-io/Facebook-dataset-samples/blob/main/facebook-posts-by-profile.csv)  

The dataset is very comprehensive: It covers a wide area of content like posts, reels, events and comments. Hashtags and engagement metrics can be analyzed. We can also examine user interactions like using likes and replies to improve virality.


Strengths:
* Contains many useful attributes for further analysis
* Has multiple post formats
* Has post date and specific time so we can see the time of day effects for a post
* Since I believe its part of the official dataset, this data should be comprehensive and useful

Weaknesses:
* Still lacks video duration and audio information, so we cannot fully study what makes a video viral
* Has quite a lot of null attributes, and date format needs to be standardised, meaning needs some data processing

```python
df = pd.read_csv('facebook-posts-by-profile.csv')
```

## DS3

[Facebook Metrics Data (DS3) - Facebook Metrics of Cosmetic Brand.csv](https://www.kaggle.com/datasets/dileeppatchaone/facebook-metrics-dataset-of-cosmetic-brand)  

This dataset comes from a cosmetics brand page and contains performance metrics for Facebook posts published over a year. It has no null values and includes about 500 posts with various engagement and metadata metrics, basically the usual likes, shares and comments etc.

Strengths:
* Also relevant to engagement virality analysis.
* Has temporal data as well
* Has a lifetime post data that might be helpful

Weaknesses:
* Not sure about its usefulness / relevance to the overarching theme which is that this is solely about cosmetics, and maybe our topic is about entertainment?
* Also does not have multimedia data
* Small sample of around 500 posts
