```python
import pandas as pd
```

```python
# Load posts and reels data
df_posts = pd.read_csv('instagram/Instagram - Posts.csv')
df_reels = pd.read_csv('instagram/Instagram - Reels.csv')
```

### for the hashtags section, we will only need to consider the hashtags column as the independent variable, as well as the engagement metrics to measure virality as the dependent variable, namely num_comments and likes. We will consider views for reels separately since theres no views column for posts.

```python
# Combine hashtags and engagement metrics
posts_subset = df_posts[['hashtags', 'num_comments', 'likes']]
reels_subset = df_reels[['hashtags', 'num_comments', 'likes', 'views', 'followers']]
df_combined = pd.concat([posts_subset, reels_subset], ignore_index=True)
df_combined
```

### We assume that NaN rows in hashtags means the post does not have any hashtags. We will now check for NaN or <0 rows for num_comments, likes, views, followers.

```python
# Function to check for problematic rows
def checkForProblematicRows(df):
    df_check = df.copy()
    
    # For views, we will start checking from row 1000 onwards (ignoring first 1000 rows because they are definitely NaN)
    views = (df_check['views'].isna() | (df_check['views'] < 0)) & (df_check.index >= 1000)
    
    # For num_comments and likes, check all rows
    comments = df_check['num_comments'].isna() | (df_check['num_comments'] < 0)
    likes = df_check['likes'].isna() | (df_check['likes'] < 0)

    # For followers, we will start checking from row 1000 onwards as well.
    followers = (df_check['followers'].isna() | (df_check['followers'] < 0)) & (df_check.index >= 1000)
    
    # Combine all conditions
    final_combined_attributes = comments | likes | views | followers
    
    # Get the rows that meet any of the conditions
    problematic_rows = df_check[final_combined_attributes]

    return problematic_rows
```

```python
problematic_rows = checkForProblematicRows(df_combined)
problematic_rows
```

### We will now eliminate the NaN values for each column.

### For NaN rows for hashtags, we will assume that the user did not include any hashtags, and so we will replace the NaN rows with empty lists instead.

### For NaN and negative rows for num_comments, likes,views, followers, we will set 0 values, assuming that NaN values indicate that there are no values respectively and negative values being impossible to have.

```python
# Fill NaN with empty lists to clean hashtags
df_combined['hashtags'] = df_combined['hashtags'].fillna('[]')

# Replace NaN and negative values in num_comments with 0
df_combined.loc[df_combined['num_comments'].isna() | (df_combined['num_comments'] < 0), 'num_comments'] = 0

# Replace NaN and negative values in likes with 0
df_combined.loc[df_combined['likes'].isna() | (df_combined['likes'] < 0), 'likes'] = 0

# For views, only replace NaN and negative values from row 1000 onwards
views = (df_combined.index >= 1000) & (df_combined['views'].isna() | (df_combined['views'] < 0))
df_combined.loc[views, 'views'] = 0

# For followers, only replace NaN and negative values from row 1000 onwards
followers = (df_combined.index >= 1000) & (df_combined['followers'].isna() | (df_combined['followers'] < 0))
df_combined.loc[followers, 'followers'] = 0
```

```python
problematic_rows = checkForProblematicRows(df_combined)
problematic_rows
```

### We will change the type of hashtags to become an actual list for easier processing

```python
import ast

print('Before:', type(df_combined['hashtags'].iloc[1]))

# Convert string representation of list to actual list objects
df_combined['hashtags_list'] = df_combined['hashtags'].apply(ast.literal_eval)

print('After:', type(df_combined['hashtags_list'].iloc[1]))
```

### Enriching by new attribute has_hashtags

simple classification:
- Empty list: False
- Non-empty lists: True

```python
# Create a category: 'With Hashtags' vs 'Without Hashtags'
df_combined['has_hashtags'] = df_combined['hashtags_list'].apply(lambda x: True if len(x) > 0 else False)
```

```python
# Drop initial hashtags column because we already have a clean version
df_combined = df_combined.drop(columns=['hashtags'])

df_combined
```

### Enriching by new attribute hashtag_count

Get the length of the list to see if the number of hashtags affect the like, comment, view count

```python
# Create the hashtag_count column
df_combined['hashtag_count'] = df_combined['hashtags_list'].apply(lambda x: len(x) if isinstance(x, list) else 0)

df_combined
```

### Hashtag existence vs engagement

```python
# Group by hashtag status and look at the average engagement
posts = df_combined.head(1000)
reels = df_combined.tail(1000)

summary_posts = posts.groupby('has_hashtags')[['likes', 'num_comments']].agg(['mean', 'median', 'count'])

summary_reels = reels.groupby('has_hashtags')[['likes', 'num_comments']].agg(['mean', 'median', 'count'])

summary_combined = df_combined.groupby('has_hashtags')[['likes', 'num_comments']].agg(['mean', 'median', 'count'])

print(summary_posts)
print('*****************************************************************')
print(summary_reels)
print('*****************************************************************')
print(summary_combined)
```

Key Findings:
1. Posts with hashtags are more common
Hashtags present: 1,303 posts (65% of total)
No hashtags: 697 posts (35% of total)
Users clearly prefer using hashtags

2. Counter-intuitive engagement patterns
For Posts (first 1000 rows):
Without hashtags: Higher average likes (5,530 vs 1,925) and comments (79.9 vs 25.5)
But median values tell a different story: median likes (59 vs 47) and comments (2 vs 1) are closer


For Reels (last 1000 rows):
Same pattern: Without hashtags perform better on average
Likes: 15,417 vs 6,465 | Comments: 96.7 vs 75.5

Combined dataset:
Without hashtags: 10,410 average likes vs 4,211 with hashtags
Comments: 88.2 vs 50.7 without hashtags

What This Really Means:
High mean but lower median for non-hashtag posts suggests that very large accounts (celebrities, major brands) often don't need hashtags
These accounts have huge follower bases, so their posts get high engagement regardless of hashtags
This skews the average upward for the "no hashtags" group

For "regular" accounts:
Posts: The same result
Reels: Very close medians (171 with hashtags vs 207 without)
Combined: Almost identical median likes (97 vs 108) and comments (4 vs 4)

Conclusion:
The data suggests that hashtags don't dramatically impact engagement for typical posts, but:
Without hashtags: You get extreme highs (from big accounts) and extreme lows
With hashtags: More consistent, predictable performance
Insight: Hashtags might be more about discoverability than engagement. They help new accounts get found, but once you have a following, they matter less.

```python
from scipy.stats import mannwhitneyu
import numpy as np

def mann_whitney_u_test(df):
    likes_with_hashtags = df[df['has_hashtags'] == True]['likes']
    likes_without_hashtags = df[df['has_hashtags'] == True]['likes']

    # Perform Mann-Whitney U test for likes
    u_stat, p_value_likes = mannwhitneyu(
        likes_with_hashtags, 
        likes_without_hashtags, 
        alternative='two-sided'
    )

    comments_with_hashtags = df[df['has_hashtags'] == True]['num_comments']
    comments_without_hashtags = df[df['has_hashtags'] == False]['num_comments']

    # Perform Mann-Whitney U test for comments
    u_stat_comments, p_value_comments = mannwhitneyu(
        comments_with_hashtags, 
        comments_without_hashtags,
        alternative='two-sided'
    )

    print(f"Likes U-statistic: {u_stat}, p-value: {p_value_likes:.4f}")
    print(f"Comments U-statistic: {u_stat_comments}, p-value: {p_value_comments:.4f}")
```

```python
print('Posts:')
mann_whitney_u_test(df_combined.head(1000))
print('***********************************************')
print('Reels:')
mann_whitney_u_test(df_combined.tail(1000))
print('***********************************************')
print('Combined:')
mann_whitney_u_test(df_combined)
```

Hashtags doesnt play much signifcance in boosting engagement

### Number of hashtags vs engagement

```python
# Correlation between hashtag count and engagement
from scipy.stats import spearmanr

def correlation_analysis(df, attribute):
    hashtag_users = df[df['has_hashtags'] == True]
    correlation, p_value = spearmanr(hashtag_users['hashtag_count'], 
                                     hashtag_users[attribute])
    print(f"Correlation between hashtag count and {attribute}: {correlation:.3f}, p={p_value:.3f}")
```

```python
correlation_analysis(df_combined, 'likes')
correlation_analysis(df_combined, 'num_comments')
correlation_analysis(df_combined.tail(1000), 'views')
```

Likes: Weak negative correlation (-0.163, p < 0.001)

Comments: Weak negative correlation (-0.127, p < 0.001)

Views: No correlation (-0.017, p = 0.658)

### Normalise reels data

```python
import pandas as pd
import numpy as np
from scipy.stats import mannwhitneyu, spearmanr
import matplotlib.pyplot as plt
import seaborn as sns

reels_data = df_combined.tail(1000).copy()

# Create normalised metrics (avoiding division by zero)
reels_data['engagement_rate'] = (reels_data['likes'] + reels_data['num_comments']) / reels_data['followers'].replace(0, np.nan) * 100
reels_data['likes_per_view'] = reels_data['likes'] / reels_data['views'].replace(0, np.nan) * 100
reels_data['comments_per_view'] = reels_data['num_comments'] / reels_data['views'].replace(0, np.nan) * 100
reels_data['likes_per_follower'] = reels_data['likes'] / reels_data['followers'].replace(0, np.nan) * 100

# Check the normalised metrics
print("Normalised Metrics Summary:")
print(reels_data[['engagement_rate', 'likes_per_view', 'comments_per_view', 'likes_per_follower']].describe())
```

```python
# Split by hashtag existence
with_hashtags_norm = reels_data[reels_data['has_hashtags'] == True]
without_hashtags_norm = reels_data[reels_data['has_hashtags'] == False]

# Define metrics to test
normalized_metrics = ['engagement_rate', 'likes_per_view', 'comments_per_view', 'likes_per_follower']

# Run Mann-Whitney U tests for each normalized metric
print("=" * 60)
print("MANN-WHITNEY U TEST - HASHTAG EXISTENCE (NORMALIZED DATA)")
print("=" * 60)

results = []
for metric in normalized_metrics:
    # Drop NaN values
    with_data = with_hashtags_norm[metric].dropna()
    without_data = without_hashtags_norm[metric].dropna()
    
    if len(with_data) > 0 and len(without_data) > 0:
        u_stat, p_value = mannwhitneyu(with_data, without_data, alternative='two-sided')
        
        # Calculate medians
        median_with = with_data.median()
        median_without = without_data.median()
        
        results.append({
            'Metric': metric,
            'U-Statistic': u_stat,
            'P-Value': p_value,
            'Significant': p_value < 0.05,
            'Median (With Hashtags)': median_with,
            'Median (Without)': median_without,
            'Difference': median_with - median_without,
            'With N': len(with_data),
            'Without N': len(without_data)
        })

results_df = pd.DataFrame(results)
print(results_df.to_string())
```

```python
# Filter to only posts with hashtags and valid normalized metrics
hashtag_users_norm = reels_data[reels_data['has_hashtags'] == True].copy()

print("\n" + "=" * 60)
print("SPEARMAN CORRELATION - HASHTAG COUNT VS NORMALIZED METRICS")
print("=" * 60)

correlation_results = []
for metric in normalized_metrics:
    # Drop NaN values
    valid_data = hashtag_users_norm[[ 'hashtag_count', metric]].dropna()
    
    if len(valid_data) > 0:
        correlation, p_value = spearmanr(valid_data['hashtag_count'], valid_data[metric])
        
        correlation_results.append({
            'Metric': metric,
            'Correlation': correlation,
            'P-Value': p_value,
            'Significant': p_value < 0.05,
            'N': len(valid_data)
        })

correlation_df = pd.DataFrame(correlation_results)
print(correlation_df.to_string())
```

### Compare popular vs niche hashtags

```python
import pandas as pd
import numpy as np
from scipy.stats import mannwhitneyu
import matplotlib.pyplot as plt
import seaborn as sns

# First, let's identify the top 20 hashtags from your data
all_hashtags = []
hashtag_lists = df_combined[df_combined['has_hashtags'] == True]['hashtags_list'].tolist()

for hashtag_list in hashtag_lists:
    if isinstance(hashtag_list, list):
        all_hashtags.extend([tag.lower().strip('#') for tag in hashtag_list])

from collections import Counter
hashtag_counts = Counter(all_hashtags)
top_20_hashtags = [tag for tag, count in hashtag_counts.most_common(20)]

print("=" * 70)
print("TOP 20 MOST COMMON HASHTAGS")
print("=" * 70)
for i, tag in enumerate(top_20_hashtags, 1):
    print(f"{i:2d}. #{tag} (used {hashtag_counts[tag]} times)")

print(f"\nTotal unique hashtags: {len(hashtag_counts)}")
print(f"Top 20 represent: {(sum(hashtag_counts[tag] for tag in top_20_hashtags) / len(all_hashtags) * 100):.1f}% of all hashtag uses")
```

```python
def contains_top_hashtags(hashtag_list, top_tags):
    """Check if post contains any top 20 hashtag"""
    if not isinstance(hashtag_list, list):
        return False
    
    hashtag_list = [tag.lower().strip('#') for tag in hashtag_list]
    return any(tag in top_tags for tag in hashtag_list)

def count_top_hashtags(hashtag_list, top_tags):
    """Count how many top 20 hashtags are used"""
    if not isinstance(hashtag_list, list):
        return 0
    
    hashtag_list = [tag.lower().strip('#') for tag in hashtag_list]
    return sum(1 for tag in hashtag_list if tag in top_tags)

# Apply to dataset
df_combined['uses_top_hashtag'] = df_combined['hashtags_list'].apply(
    lambda x: contains_top_hashtags(x, top_20_hashtags)
)

df_combined['top_hashtag_count'] = df_combined['hashtags_list'].apply(
    lambda x: count_top_hashtags(x, top_20_hashtags)
)

df_combined['top_hashtag_pct'] = df_combined.apply(
    lambda row: (row['top_hashtag_count'] / len(row['hashtags_list']) * 100) 
    if isinstance(row['hashtags_list'], list) and len(row['hashtags_list']) > 0 else 0, 
    axis=1
)

print("\n" + "=" * 70)
print("POSTS USING TOP 20 HASHTAGS")
print("=" * 70)
print(f"Posts using at least one top 20 hashtag: {df_combined['uses_top_hashtag'].sum()} ({df_combined['uses_top_hashtag'].mean()*100:.1f}%)")
print(f"Posts with NO top 20 hashtags: {(~df_combined['uses_top_hashtag']).sum()} ({(~df_combined['uses_top_hashtag']).mean()*100:.1f}%)")
```

```python
# Separate datasets
posts_data = df_combined.head(1000).copy()
reels_data = df_combined.tail(1000).copy()

# Calculate normalized metrics for reels
reels_data['engagement_rate'] = (reels_data['likes'] + reels_data['num_comments']) / reels_data['followers'].replace(0, np.nan) * 100
reels_data['likes_per_view'] = reels_data['likes'] / reels_data['views'].replace(0, np.nan) * 100

print("\n" + "=" * 70)
print("PERFORMANCE COMPARISON: TOP HASHTAGS vs OTHERS")
print("=" * 70)

# For Posts (raw numbers)
posts_top = posts_data[posts_data['uses_top_hashtag'] == True]
posts_other = posts_data[posts_data['uses_top_hashtag'] == False]

print("\n--- POSTS (Raw Numbers) ---")
print(f"{'Metric':<20} {'With Top Hashtags':>20} {'Without Top':>20} {'Difference':>15}")
print("-" * 75)

for metric in ['likes', 'num_comments']:
    top_median = posts_top[metric].median()
    other_median = posts_other[metric].median()
    diff_pct = ((top_median - other_median) / other_median * 100) if other_median > 0 else float('inf')
    
    print(f"{metric:<20} {top_median:>20.1f} {other_median:>20.1f} {diff_pct:>+14.1f}%")
    
    # Statistical test
    u_stat, p_value = mannwhitneyu(posts_top[metric], posts_other[metric])
    print(f"{'p-value':<20} {p_value:>20.4f} {'Significant' if p_value < 0.05 else 'Not significant':>20}")

# For Reels (normalized)
reels_top = reels_data[reels_data['uses_top_hashtag'] == True]
reels_other = reels_data[reels_data['uses_top_hashtag'] == False]

print("\n--- REELS (Normalized) ---")
print(f"{'Metric':<20} {'With Top Hashtags':>20} {'Without Top':>20} {'Difference':>15}")
print("-" * 75)

for metric in ['engagement_rate', 'likes_per_view']:
    if metric in reels_top.columns and metric in reels_other.columns:
        top_median = reels_top[metric].median()
        other_median = reels_other[metric].median()
        diff_pct = ((top_median - other_median) / other_median * 100) if other_median > 0 else float('inf')
        
        print(f"{metric:<20} {top_median:>20.2f}% {other_median:>20.2f}% {diff_pct:>+14.1f}%")
        
        # Statistical test
        u_stat, p_value = mannwhitneyu(reels_top[metric].dropna(), reels_other[metric].dropna())
        print(f"{'p-value':<20} {p_value:>20.4f} {'Significant' if p_value < 0.05 else 'Not significant':>20}")
```
